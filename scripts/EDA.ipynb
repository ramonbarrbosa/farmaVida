# EDA (Análise Exploratória de Dados) #

### Descritivos ###

Na maioria dos casos, os dados reais apresentam problemas como valores ausentes, dados duplicados ou inconsistentes. 

A qualidade dos dados é um fator crucial para o sucesso da análise, e a EDA é especialmente útil para identificar e resolver essas questões.

A limpeza de dados envolve várias etapas, como:

- Tratamento de dados ausentes;
- Remoção de duplicatas;
- Correção de inconsistências;
- Conversão de tipos de dados.

### Refinamento ###

**Checar as Chaves Primárias (Duplicadas nas Dimensões)**

```python
# Já vimos que a linha inteira não é duplicada, mas vale conferir se os IDs são únicos:
# Devem todos retornar True (sem IDs repetidos)
 
print(f'Dimensão produtos: {df_produtos['cod_produto'].is_unique}') 
print(f'Dimensão clientes: {df_clientes['cod_cliente'].is_unique}') 
print(f'Dimensão vendedores: {df_vendedores['cod_vendedor'].is_unique}') 
print(f'Dimensão lojas: {df_lojas['cod_loja'].is_unique}')
```

**Investigando as Chaves Orfãs em vendas**

```python
# Vale rodar um filtro rápido para ver o tamanho do problema: 
# Lembra que no describe() o cod_produto em vendas chegou a 999, mas a tabela de produtos só tem 134?
# >>> (Uma venda apontando para um produto inexistente no catálogo) <<<

orfaos_prod = df_vendas[~df_vendas['cod_produto'].isin(df_produtos['cod_produto'])] 
print(f"Linhas de vendas com produto não cadastrado: {len(orfaos_prod)}")
```

```python
# Descobre quais códigos vendidos não existem no cadastro de produtos: 
codigos_nao_cadastrados = set(df_vendas['cod_produto']) - set(df_produtos['cod_produto']) 
print(f"Códigos de produtos em vendas que não existem na dimensão produtos: {codigos_nao_cadastrados}") 

# Quantidade de vendas afetadas: 
qtd_linhas_afetadas = df_vendas['cod_produto'].isin(codigos_nao_cadastrados).sum() 
print(f"Total de registros de vendas afetados: {qtd_linhas_afetadas}")
```

## Conclusão ##

### 1. Qualidade e Integridade da Base

Volumetria: 10 tabelas auditadas a partir dos arquivos .csv, cobrindo mais de 223 mil registros de vendas
(período de 2023 a 2024).

Duplicatas: Nenhuma duplicata encontrada a nível de linha nas 10 tabelas analisadas.
Unicidade das Dimensões: As chaves primárias (cod_produto, cod_cliente, cod_vendedor, cod_loja) são
100% únicas (is_unique = True), garantindo relacionamentos 1:N no Power BI sem risco de cardinalidade muitos- para-muitos.

### 2. Validação das Regras de Negócio e Nulos

segmento_pj (clientes): 97% dos registros são nulos por corresponderem a clientes Pessoa Física (PF). Os
registros foram preservados para manter a totalidade da base.

data_fim (comissões) e data_desligamento (vendedores): Os valores nulos representam registros ativos
(comissões vigentes e colaboradores ativos).

data_encerramento (lojas): Nulos indicam lojas em plena operação comercial.
### 3. Ponto Crítico Identificado para o Power BI (Chaves Órfãs)

Foram detectadas 60 linhas de vendas associadas ao cod_produto = 999, código inexistente na dimensão
produtos (que possui itens de 1 a 134).

Ação no Power Query: Tratar essa chave criando uma linha/categoria para "Produto Não Cadastrado / Outros",
evitando que o Power BI gere agrupamentos em branco (Blank) nos visuais de produto e categoria.
