# EDA (Análise Exploratória de Dados) #

### Descritivos ###

Na maioria dos casos, os dados reais apresentam problemas como valores ausentes, dados duplicados ou inconsistentes. 

A qualidade dos dados é um fator crucial para o sucesso da análise, e a EDA é especialmente útil para identificar e resolver essas questões.

A limpeza de dados envolve várias etapas, como:

- Tratamento de dados ausentes;
- Remoção de duplicatas;
- Correção de inconsistências;
- Conversão de tipos de dados.

### SQLite ###

```python
import sqlite3 as sql
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Conexao com o banco de dados SQLite
def connect_db():
    try:
        # Modo rw garante que o arquivo farmavida.db já existe na pasta atual
        con = sql.connect('file:../database/farmavida.db?mode=rw', uri=True)
        print('Conexão com o banco de dados estabelecida com sucesso!')
        return con
    except sql.OperationalError:
        print('Erro: O arquivo "farmavida.db" não foi encontrado no diretório.')
        return None
    except Exception as e:
        print(f'Erro ao conectar ao banco de dados: {e}')
        return None

# Chamada da função
con = connect_db()
cur = con.cursor()
```

```python
table = [
    'cidades', 'clientes', 'comissoes', 
    'custo_produto','devolucoes', 'lojas', 
    'metas_vendedor', 'produtos', 'vendas','vendedores'
]
```

```python
# Carregando os dados das tabelas para um DataFrame
for f in table:
    query = cur.execute('select * from {}'.format(f))
    df = pd.DataFrame(query.fetchall(), columns=[i[0] for i in query.description])
    print(f'\n--- Tabela {f} ---')
    display(df.head(5))
```

```python
# Fechando a conexão com o banco de dados
con.close()
```
